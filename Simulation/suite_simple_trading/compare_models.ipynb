```python
### Global Imports
import os
import pandas as pd
```

```python
### Import Checks
```

```python
### Global Variables
DATA_SAVE_PATH = "../../models/used_data"
DAYS_PER_EPISODE = 3
TRAIN_TEST_SPLIT_FRACTION = 0.2  # Amount of data you reserve for testing.
BUFFER_DAYS = 3  # Amount of days that needs to be between the test episodes.
```

```python
from Simulation.suite_simple_trading.data_splitting import get_or_create_train_test_split
from Simulation.suite_simple_trading.pre_processing import clean_data

### Getting data

raw_data_path = '../../data/2025_minute.csv'
cleaned_data_cache_path = '../../data/2025_minute_cleaned.pkl'

if os.path.exists(cleaned_data_cache_path):
    print(f"Loading cached cleaned data from: {cleaned_data_cache_path}")
    cleaned_df = pd.read_pickle(cleaned_data_cache_path)
else:
    print("No cached data found. Running the full cleaning process...")
    raw_df = pd.read_csv(raw_data_path, sep=';')
    cleaned_df = clean_data(raw_df)
    print(f"Saving cleaned data to cache: {cleaned_data_cache_path}")
    cleaned_df.to_pickle(cleaned_data_cache_path)

all_data = cleaned_df[['Datetime', 'Imbalance Price']]

train_df, test_df, nr_of_episodes = get_or_create_train_test_split(
    all_data=all_data,
    save_path=DATA_SAVE_PATH,
    days_per_episode=DAYS_PER_EPISODE,
    test_fraction=TRAIN_TEST_SPLIT_FRACTION,
    buffer_days=BUFFER_DAYS
)
```

### Training agents

```python
### The model (Environment) used in all agents

from Simulation.suite_simple_trading.model import ExtendedBatteryEnv

train_env = ExtendedBatteryEnv(
    battery_capacity_mwh=10.0,
    charge_discharge_rate_mw=5.0,
    all_data=train_df,
    days_per_episode=DAYS_PER_EPISODE
)
```

#### PPO

```python
from Simulation.suite_simple_trading.agent_trainer import EpisodeRewardCallback
from sb3_contrib import MaskablePPO
from sb3_contrib.common.maskable.policies import MaskableActorCriticPolicy
from Simulation.suite_simple_trading.observation_wrappers import RobustScalingWrapper

### PPO
PPO_TRAINING_TIMESTEPS = 100_000
MODEL_SAVE_PATH_PPO = "../../models"
# Scaling features
train_env_scaled = RobustScalingWrapper(train_env)

ppo_agent = MaskablePPO(MaskableActorCriticPolicy, train_env_scaled)
reward_callback = EpisodeRewardCallback(verbose=1)
print(f"\n--- Starting Training for {PPO_TRAINING_TIMESTEPS} Timesteps ---")
ppo_agent.learn(total_timesteps=PPO_TRAINING_TIMESTEPS, progress_bar=True, callback=reward_callback)
print("--- Training Complete ---")
ppo_agent.save(MODEL_SAVE_PATH_PPO)
print(f"Trained model saved to: {MODEL_SAVE_PATH_PPO}.zip")

ppo_reward_per_episode = reward_callback.episode_rewards
```

#### DQN

```python
from Simulation.DRL_implementations.Masked_DQN import MaskedDoubleDQN

### DQN
DQN_TRAINING_TIMESTEPS = 100_000
dqn_agent = MaskedDoubleDQN(train_env, verbose=0, eps_decay=30000, train_every=1, b_size=32, update_target_every=5000)
dqn_agent.learn(DQN_TRAINING_TIMESTEPS)
```

#### EVALUATING

```python
from Simulation.suite_simple_trading.policy import PPOAgentDecisionMaker
from Simulation.suite_simple_trading.simulation import run_evaluation

### Testing
test_env = ExtendedBatteryEnv(
    battery_capacity_mwh=10.0,
    charge_discharge_rate_mw=5.0,
    all_data=test_df,
    days_per_episode=DAYS_PER_EPISODE
)
test_env_scaled = RobustScalingWrapper(test_env)
```

#### PPO

```python
decision_maker_PPO = PPOAgentDecisionMaker(ppo_agent, test_env)

result_ppo = run_evaluation(test_env_scaled, decision_maker_PPO, number_of_episodes=nr_of_episodes)
print(f"reward ppo: {sum(result_ppo["rewards"])}")
```

#### DQN

```python
decision_maker_DQN = PPOAgentDecisionMaker(dqn_agent, test_env)
result_dqn = run_evaluation(test_env_scaled, decision_maker_DQN, number_of_episodes=nr_of_episodes)
print(f"reward dqn: {sum(result_dqn["rewards"])}")
```
